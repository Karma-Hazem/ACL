```python

import os
import glob

# Set the dataset folder path from Kaggle's Input section
data_folder = "/kaggle/input/datasets/rohanrao/formula-1-world-championship-1950-2020"

# Find all CSV files inside the dataset folder
csv_files = glob.glob(
    os.path.join(data_folder, "**", "*.csv"),
    recursive=True
)

# Load all CSV files into a dictionary
data = {
    os.path.splitext(os.path.basename(file))[0]: pd.read_csv(
        file, na_values=[r"\N"]
    )
    for file in csv_files
}

# Display the loaded tables
print("Number of CSV files loaded:", len(data))

for name, df in data.items():
    print(f"{name}: {df.shape}")
```

```python
for name, df in data.items(): #checking first 5 rows of each table fel dataset 
    print(f"\n{'=' * 50}")
    print(f"Table: {name}")
    print(f"Shape: {df.shape}")
    display(df.head())
```

```python
#checking the datatypes
for name, df in data.items():
    print(f"\n--- {name} ---")
    print(data[name].dtypes)
```

```python
#checking missing values
for name, df in data.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0].sort_values(ascending=False)

    print(f"\n--- {name} ---")

    if missing.empty:
        print("No missing values found.")
    else:
        missing_table = pd.DataFrame({
            "Missing Count": missing,
            "Missing Percentage (%)": (
                missing / len(df) * 100
            ).round(2)
        })
        display(missing_table)
```

```python
#checking duplicayes
for name, df in data.items():
    print(
        f"{name}: {df.duplicated().sum()}  duplicated rows"
    )
```

```python
# Check whether primary IDs are duplicated
primary_keys = {
    "circuits": ["circuitId"],
    "constructors": ["constructorId"],
    "drivers": ["driverId"],
    "races": ["raceId"],
    "results": ["resultId"],
    "qualifying": ["qualifyId"],
    "status": ["statusId"]
}

for table, keys in primary_keys.items():
    if table in data and all(k in data[table].columns for k in keys):
        df = data[table]
        duplicates = df.duplicated(subset=keys).sum()

        print(f"{table}: {duplicates} duplicate IDs")
```
